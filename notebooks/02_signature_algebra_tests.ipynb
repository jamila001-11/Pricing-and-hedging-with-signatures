```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import numpy as np

from models import simulate_shifted_bergomi
from signatures import compute_signatures, generate_words, projection, shuffle_product, word_to_index, zeros_tensor
from signatures.bergomi_kernel import compute_sigma_t
```

```python
M = 3
words = generate_words(M)
word_idx = word_to_index(words)

a = zeros_tensor(M)
b = zeros_tensor(M)
a[word_idx[(1,)]] = 1
b[word_idx[(2,)]] = 1

shuffle = shuffle_product(a, b, words, word_idx, M)
proj = projection(shuffle, 2, words, word_idx, M)

print('nombre de mots:', len(words))
print('shuffle(1,2) coeffs:', {w: shuffle[word_idx[w]] for w in words if shuffle[word_idx[w]] != 0})
print('projection lettre 2:', {w: proj[word_idx[w]] for w in words if proj[word_idx[w]] != 0})
```

```python
sigma_t = compute_sigma_t(t=0.01, varsigma=1.6, f0=0.0625, epsilon=1/52, H=0.1, M=M, words=words, word_idx=word_idx)
print('sigma_t mot vide:', sigma_t[word_idx[()]])
print('norme sigma_t:', np.linalg.norm(sigma_t))
```
