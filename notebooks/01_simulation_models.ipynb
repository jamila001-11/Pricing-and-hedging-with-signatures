```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import numpy as np

from models import (
    compute_payoff_asian,
    compute_payoff_europeen,
    compute_payoff_lookback,
    simulate_heston,
    simulate_shifted_bergomi,
)
```

```python
S0 = 1.0
V0 = 0.0625
T = 0.5
J = 126
I = 2_000
K = 1.0
seed = 42
```

```python
S_h, V_h, dS_h, sigma_h, states_h = simulate_heston(
    S0=S0, V0=V0, kappa=4, theta=V0, eta=0.7, rho=-0.7, T=T, J=J, I=I, seed=seed
)

payoffs_h = {
    'european': compute_payoff_europeen(S_h, K),
    'asian': compute_payoff_asian(S_h, K),
    'lookback': compute_payoff_lookback(S_h),
}

print(S_h.shape, V_h.shape, dS_h.shape, states_h.shape)
{k: float(v.mean()) for k, v in payoffs_h.items()}
```

```python
S_b, dS_b, sigma_b, states_b, W_hat_b = simulate_shifted_bergomi(
    S0=S0, varsigma=1.6, f0=V0, epsilon=1/52, H=0.1, rho=-0.7, T=1/12, J=J, I=I, seed=seed
)

print(S_b.shape, dS_b.shape, sigma_b.shape, states_b.shape, W_hat_b.shape)
print('European payoff mean:', compute_payoff_europeen(S_b, K).mean())
```
